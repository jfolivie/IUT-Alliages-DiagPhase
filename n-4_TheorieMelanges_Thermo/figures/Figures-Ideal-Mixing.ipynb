```python
import matplotlib.pyplot as plt
from matplotlib import rc
import matplotlib.colors as mcolors
import matplotlib as mpl 
import numpy as np

_dpi = 200


def colormap(num_colors, name = 'rainbow') :
    cmap = mpl.cm.get_cmap(name = name, lut = num_colors)
    return cmap(range(num_colors))

rc('text', usetex=True)
rc('font', family='serif', size = 25)

_R = 8.314
```

```python
def DSm(T, x) :
    return -((1-x) * np.log(1-x) + x * np.log(x))

def DHm(T, x, chiAB) :
    return chiAB * T * x * (1-x)


def DGm(T, x, chiAB) : 
    return DHm(T, x, chiAB) - T * DSm(T, x)

```

# Solution idéale

```python
x = np.linspace(0, 1, num = 10001)

fig = plt.figure(figsize = (12,8))
ax = fig.subplots()

ax.plot(x, DGm(1, x, 0), color = 'black')

ax.set_xlim(0,1)

ax.set_ylabel("$\\frac{\Delta G_m}{R T n^s}$ [$\emptyset$]")
ax.set_xlabel("Fraction molaire $x_B$")
ax.grid()

fig.tight_layout()

fig.savefig("Fig04a", dpi = _dpi)
```

```python
modele = lambda x, muA, muB, T, W : (1-x) * muA + x * muB + (1-x) * _R * T * np.log(1-x) + _R * T * x * np.log(x) + W * x * (1-x)

get_muB = lambda x, muB, T, W  : muB + _R * T * np.log(x) + W * x**2
get_muA = lambda x, muA, T, W  : muA + _R * T * np.log(1-x) + W * (1-x)**2

tangent = lambda x, x0, muA, muB, T, W : get_muB(x0, muB, T, W) * x + get_muA(x0, muA, T, W) * (1-x)


def corde(x, xa, xb, muA, muB, T, W, values = False) : 
    Ga = modele(xa, muA, muB, T, W)
    Gb = modele(xb, muA, muB, T, W)
    
    xmask = np.ma.masked_outside(x, xa, xb)
    if not values : 
        return (Gb - Ga) / (xb - xa) * (xmask - xa) + Ga
    else : 
        return (Gb - Ga) / (xb - xa) * (xmask - xa) + Ga, (xa, xb), (Ga, Gb)
    
muA = 10000
muB = 20000
T = 298
W = -5000

x_cords = ((0.05, 0.3), (0.4, 0.7), (0.6, 0.9), (0.8, 0.99))
pen_matplotlib=colormap(len(x_cords), name = "coolwarm")


x = np.linspace(0, 1, num = 500)

fig = plt.figure(figsize=(12,8))
ax = fig.subplots()
ax.plot(x, modele(x, muA, muB, T, W), color = 'black')


for n in range(len(x_cords)) : 
    cord, xab, Gab = corde(x, x_cords[n][0], x_cords[n][1], muA, muB, T, W, values = True)
    ax.plot(x, cord, color = pen_matplotlib[n])
    ax.plot(xab, Gab, color = pen_matplotlib[n], marker = 'o', ls = 'none')

print(np.unique(np.array(x_cords)))

ax.grid()
ax.set_xlim(0,1)

ax.set_xticks(np.ravel(np.array(x_cords)))
xticks = np.unique([('$x_{}^\\alpha$'.format(i+1), '$x_{}^\\beta$'.format(i+1)) for i in range(len(x_cords))])
print(xticks)
ax.set_xticklabels(xticks)
ax.set_yticks([])

ax.set_ylabel("Enthalpie libre molaire $\\bar{G}$")
ax.set_xlabel("Fraction molaire $x$")

fig.tight_layout()
fig.savefig("Fig05b", dpi = _dpi)
```

```python
import matplotlib.pyplot as plt
from matplotlib import rc
import matplotlib.colors as mcolors
import matplotlib as mpl 
import numpy as np

_dpi = 100


def colormap(num_colors, name = 'rainbow') :
    cmap = mpl.cm.get_cmap(name = name, lut = num_colors)
    return cmap(range(num_colors))

rc('text', usetex=True)
rc('font', family='serif', size = 25)

_R = 8.314
```

```python
def L(T, DfusH, Tfus) : 
    return DfusH/_R * (1/T - 1/Tfus)

#def xB_l(T, DfusHA, DfusHB, TfusA, TfusB) : 
#    return (np.exp(L(T, DfusHA, TfusA)) - 1) / (np.exp(L(T, DfusHB, TfusB)) - np.exp(L(T, DfusHA, TfusA)))

#def xB_s(T, DfusHA, DfusHB, TfusA, TfusB) : 
#    return (np.exp(L(T, DfusHA, TfusA)) - 1) / (np.exp(L(T, DfusHA, TfusA)) - np.exp(L(T, DfusHB, TfusB))) * np.exp(L(T, DfusHB, TfusB))

def xB_l(T, DfusHA, DfusHB, TfusA, TfusB) : 
    return (np.exp(L(T, DfusHA, TfusA)) - 1) / (np.exp(L(T, DfusHA, TfusA)) - np.exp(L(T, DfusHB, TfusB)))

def xB_s(T, DfusHA, DfusHB, TfusA, TfusB) : 
    return (np.exp(L(T, DfusHA, TfusA)) - 1) / (np.exp(L(T, DfusHA, TfusA)) - np.exp(L(T, DfusHB, TfusB))) * np.exp(L(T, DfusHB, TfusB))
```

# Diagramme Cu-Ni, mélange homogène

https://lelementarium.fr/element-fiche/nickel/

https://lelementarium.fr/element-fiche/cuivre/

```python
element = 'Ni'

TfusA = 1353.4#1358
TfusB = 1728
DfusHA = 13.26 * 10**3 #* 3.0
DfusHB = 17.48 * 10**3 #* 3.0

#TfusA = 1083.4 + 273.15
#TfusB = 1453 + 273.15
#DfusHA = 13 * 10**3 * 1.5
#DfusHB = 17.6 * 10**3 * 0.9
#DfusHA = 17.48 * 10**3

x_exp, Tliq_exp, Tsol_exp = np.loadtxt("Fig02a.csv", delimiter = ",", skiprows = 1).T
Tliq_exp += 273.15
Tsol_exp += 273.15

#print(Tliq_exp, Tsol_exp)

T = np.linspace(TfusA, TfusB, num = 1000)


fig = plt.figure(figsize = (12,8))
ax = fig.subplots()

ax.plot(xB_l(T, DfusHA, DfusHB, TfusA, TfusB), T, label = "liquidus", color = 'red')
ax.plot(xB_s(T, DfusHA, DfusHB, TfusA, TfusB), T, label = "solidus", color = 'blue')

ax.plot(x_exp, Tliq_exp, label = "liquidus exp.", color = 'red', ls = 'none', marker = 'o')
ax.plot(x_exp, Tsol_exp, label = "solidus exp.", color = 'blue', ls = 'none', marker = 'o')

#ax.set_xlim(0,1)

ax.set_ylabel("Température [$K$]")
ax.set_xlabel("Fraction molaire $x_{" + "{}".format(element)+ "}$")
ax.grid()

ax.set_xlim(0,1)

fig.legend()

fig.tight_layout()

fig.savefig("Fig02a", dpi = _dpi)
```

```python
#TfusA = TfusA
#TfusB = np.linspace(0.5 * TfusA, 1.5 * TfusA, num = 6)

TfusA = 1358
TfusB = 1728
DfusHA = 13.26 * 10**3
DfusHB = np.linspace(0.5 * DfusHA, 2.0 * DfusHA, num = 11)
#DfusHA = 17.48 * 10**3

pen_matplotlib=colormap(len(DfusHB), name = "coolwarm")

T = np.linspace(TfusA, TfusB, num = 1000)


fig = plt.figure(figsize = (12,8))
ax = fig.subplots()

#colors
cmap = plt.get_cmap('jet', len(DfusHB)) 

for n in range(len(DfusHB)) : 
    
    ax.plot(xB_l(T, DfusHA, DfusHB[n], TfusA, TfusB), T, color = pen_matplotlib[n])
    ax.plot(xB_s(T, DfusHA, DfusHB[n], TfusA, TfusB), T, color = pen_matplotlib[n], ls = "dotted")

ax.set_ylabel("Température [$K$]")
ax.set_xlabel("Fraction molaire $x_B$")
ax.set_xlim(0,1)
ax.grid()

# Create a colorbar with discrete color labels
bounds = np.arange(len(pen_matplotlib)+1) - 0.5
cmap= mcolors.ListedColormap(pen_matplotlib)
sm = plt.cm.ScalarMappable(cmap=cmap)
sm.set_array([])  # This step is necessary to create a mappable
cbar = plt.colorbar(sm, ticks=np.arange(len(pen_matplotlib)), boundaries = bounds)
cbar.set_ticklabels(np.round(DfusHB/DfusHA, 1))
cbar.set_label('$\Delta_{fus} H^{\circ, \star}_B / \Delta_{fus} H^{\circ, \star}_A$')

fig.tight_layout()
fig.savefig("Fig02d", dpi = _dpi)
```

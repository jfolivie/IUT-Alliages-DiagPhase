```python
import numpy as np
import matplotlib.pyplot as plt

_dtype = np.float32
```

```python

w = 0.1 * np.arange(0.0, 11, 1)
T1 = np.array([321, 316, 308, 302, 293, 284, 274, 263, 251, 272, 288], dtype = _dtype)

T2 = 250 * np.ones(11, dtype = _dtype)
T2[0] = np.nan
T2[-1] = np.nan

fig = plt.figure(figsize = (12, 8))
ax = fig.subplots()
ax.plot(w, T1, color = 'red', marker = 'o', ls = 'none')
ax.plot(w, T2, color = 'black', marker = 'o', ls = 'none')

ax.set_xlim(0,1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import rc

_dpi = 100


def colormap(num_colors, name = 'rainbow') :
    cmap = mpl.cm.get_cmap(name = name, lut = num_colors)
    return cmap(range(num_colors))

#(10,10), size = 25

rc('text', usetex=True)
rc('font', family='serif', size = 12)
```

```python
_NA = 6.02 * 10**(23) #/mol

T1 = 25 
T2 = 912
T3 = 1394
T4 = 1535

alpha = (2.866,2.89, T1, T2)
gamma = (3.56, 3.63, T2, T3)
delta = (2.92, 2.922, T2, T3)

def get_a(T, a1, a2, T1, T2) : 
    return (a2-a1) / (T2-T1) * (T - T1) + a1

def get_V(a, Z) : 
    return (a*10**(-8))**3 / Z * _NA

T = np.linspace(T1+1, T4, num = 1000)
mask1 = (T > T1) & (T < T2)
mask2 = (T > T2) & (T < T3)
mask3 = (T > T3)

a = np.zeros(len(T))
a[mask1] = get_a(T[mask1], *alpha)
a[mask2] = get_a(T[mask2], *gamma)
a[mask3] = get_a(T[mask3], *delta)

fig = plt.figure()
ax = fig.subplots()
ax.plot(T, a, color = 'black')

V = np.zeros(len(T))
V[mask1] = get_V(a[mask1], 2)
V[mask2] = get_V(a[mask2], 4)
V[mask3] = get_V(a[mask3], 2)


fig = plt.figure()
ax = fig.subplots()
ax.plot(T, V)

print(get_a(1394, *alpha))
print(get_a(T4, *alpha))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import rc
from scipy.optimize import curve_fit
import matplotlib.colors as mcolors

_dpi = 200


def colormap(num_colors, name = 'rainbow') :
    cmap = mpl.cm.get_cmap(name = name, lut = num_colors)
    return cmap(range(num_colors))

#(10,10), size = 25

rc('text', usetex=False)
rc('font', family='serif', size = 20)
```

```python
name = "Fig01-refroidissement"

T0 = 2
Tinfty = 1
tau = 1

tmin = 0
tmax = 5

T = lambda t : Tinfty + (T0-Tinfty) * np.exp(-t / tau)
Ti = lambda t : Tinfty + (T0-Tinfty) * (1 - t/tau)

t = np.linspace(tmin, tmax, num = 101)
taxis = np.arange(tmin, tmax+1)
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
ax.plot(t, T(t), color = 'black')
ax.plot(t, Ti(t), color = 'red')

ax.set_ylim(min(T0, Tinfty),max(T0, Tinfty))
ax.set_xlim(0,5)

ax.set_xticks(taxis)
ax.set_xticklabels(["{}$\\tau$".format(i) for i in taxis])

ax.set_yticks([T0, Tinfty])
ax.set_yticklabels(["$T_0$","$T_\\infty$"])

ax.set_xlabel("Temps")
ax.set_ylabel("Température de l'échantillon")

ax.grid()

fig.savefig(name + ".pdf", dpi = _dpi)
```

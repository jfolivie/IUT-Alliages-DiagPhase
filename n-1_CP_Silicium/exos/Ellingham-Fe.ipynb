```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib import rc

rc('lines', lw=2, c='r')
rc('font', size=14)
#rc('xticks', size=12)
```

```python
Tmin = 800
Tmax = 1200





ellingham = dict(FetoFeO = lambda T : -538 + 0.14 * T,
                FetoFe3O4 = lambda T : -551 + 0.15 * T,
                FetoFe2O3 = lambda T : -543 + 0.17 * T,
                CtoCO = lambda T : -270 - 0.11 * T,
                CtoCO2 = lambda T : -396 + 0.0 * T)
T = np.arange(Tmin, Tmax, 1)

colors = ('blue', 'green', 'red', 'black', 'black')
colors = dict(zip(ellingham.keys(),colors))

ls = ('solid', 'solid', 'solid', 'dashed', 'dotted')
ls = dict(zip(ellingham.keys(),ls))
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
for key in ellingham.keys() :
    ax.plot(T, ellingham[key](T), 
            color = colors[key], label = key.replace('to', ' <=> '), ls = ls[key])
    
ax.set_xlim(Tmin, Tmax)
ax.legend()
ax.grid()

ax.set_xlabel("$T$ [$K$]")
ax.set_ylabel("$\Delta_r G^\circ (T)$ [$K$]")

fig.savefig("../figures/Fe-C-O.pdf", dpi = 100)
```

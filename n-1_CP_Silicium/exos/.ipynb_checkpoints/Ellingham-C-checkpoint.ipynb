```python
import numpy as np
import matplotlib.pyplot as plt

_dtype_f = np.float32
```

```python
#Parameters
Tmin = 298
Tmax = 1200
dT = 1

T = np.arange(Tmin, Tmax, dT, dtype = _dtype_f)

#Parameters
DrH = dict(CtCO = -110.5, 
          COtCO2 = -282.8)
DrS = dict(CtCO = 89.5*10**(-3), 
          COtCO2 = -86.6*10**(-3))

colors = dict(zip(tuple(DrH.keys()), ('blue', 'green')))
```

```python
DrG = lambda T, DrH, DrS : DrH - T * DrS
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.subplots()

for key in DrH.keys() : 
    ax.plot(T, DrG(T,DrH[key],DrS[key]), 
            label = key, color = colors[key])

ax.plot(T, DrG(T, DrH["COtCO2"] - DrH["CtCO"]) , color = 'gray')


ax.set_xlabel("$T$ [$K$]")
ax.set_ylabel("$\Delta_r G^\circ (T)$ [$kJ \cdot mol^{-1}$]")

ax.legend()
```

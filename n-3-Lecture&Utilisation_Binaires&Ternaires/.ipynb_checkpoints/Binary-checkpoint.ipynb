```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def T(t, Tinf = 0, T0 = 0, Cp = 1, R = 1, Tliq = 0, Tsol = 0) :
    
    Tout = Tinf - (T0 -Tinf) * t / Cp / R
    
    return 

T0 = 1
Tinf = T0/4

Tliq = T0*3/4
Tsol = T0/2

t = np.linspace(0, 500, num = 501)

plt.plot(t, T(t, Tinf = Tinf, T0 = T0, Tliq = Tliq, Tsol = Tsol))
```
